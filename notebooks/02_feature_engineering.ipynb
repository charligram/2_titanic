# 3. Import data

```python
df = pd.read_csv(PROJECT_ROOT / 'data' / 'clean' / 'clean_data.csv')
```

```python
df.info()
```

# 4. Feature engineering

## 4.1 PassengerId

El id usualmente no da mucha información, primero veamos cuantas categorías tenemos y luego eliminaremos de ser necesario

```python
df['PassengerId'].nunique()
```

Todos son diferentes, eliminar

```python
df.drop(columns=['PassengerId'], inplace=True)
```

## 4.2 Name

Al igual que con PassengerId el nombre del pasajero puede no tener que ver con su sobrevivencia

```python
df['Name'].nunique()
```

```python
df.drop(columns=['Name'], inplace=True)
```

## 4.3 Sex

Como es un str lo vamos a pasar a binario

```python
df['Sex'] = df['Sex'].map({'female': 0, 'male': 1})
```

```python
df.info()
```

```python
df.head()
```

## 4.4 Ticket

El código de ticket tampoco nos va a revelar mucha información, consideremos eliminarlo

```python
df['Ticket'].nunique()
```

```python
df.drop(columns=['Ticket'], inplace=True)
```

```python
df.info()
```

## 4.5 Embarked

Para el caso de Embarked, como tenemos pocas categorías, obtenemos su info a través de one-hot-encoding

```python
# dtype=int para que convierta los bool en 0-1
df = pd.get_dummies(df, dtype=int)
```

```python
df.info()
```

```python
df.head()
```

# 5. Save

```python
df.to_csv(PROJECT_ROOT / 'data' / 'processed' / 'train_data.csv', index=False)
```
